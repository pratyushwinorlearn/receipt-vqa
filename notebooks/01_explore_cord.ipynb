# Explore CORD
Look at raw annotations before trusting the QA builder. Run `make data` first.

```python
import json
from collections import Counter

from PIL import Image

from receiptqa.utils.io import read_jsonl

rows = read_jsonl("../data/raw/cord/train.jsonl")
print(len(rows), "train receipts")
print(json.dumps(rows[0]["gt_parse"], indent=2)[:1500])
```

```python
# which fields exist, and how often?
c = Counter()
for r in rows:
    for sec, val in r["gt_parse"].items():
        if isinstance(val, dict):
            c.update(f"{sec}.{k}" for k in val)
        else:
            c[sec] += 1
c.most_common(30)
```

```python
# is `menu` ever a dict instead of a list?
sum(isinstance(r["gt_parse"].get("menu"), dict) for r in rows)
```
