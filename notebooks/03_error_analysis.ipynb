# Error analysis
Inspect the worst predictions. Set `PRED` to a predictions jsonl in outputs/results/.

```python
from collections import Counter

from PIL import Image

from receiptqa.eval.error_analysis import tag_failure
from receiptqa.utils.io import read_jsonl

PRED = "../outputs/results/REPLACE_ME.jsonl"
recs = read_jsonl(PRED)
for r in recs:
    r["tag"] = tag_failure(r)
Counter(r["tag"] for r in recs).most_common()
```

```python
fails = sorted((r for r in recs if r["tag"] != "correct"), key=lambda r: r["anls"])
for r in fails[:5]:
    print(r["qtype"], r["tag"], "|", r["question"], "| gold:", r["answer"], "| pred:", r["prediction"])
    display(Image.open("../" + r["image_path"]).resize((400, 600)))
```
